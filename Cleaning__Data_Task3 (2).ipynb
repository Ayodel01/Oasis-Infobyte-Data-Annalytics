```python
import pandas as pd
import numpy as np

data = {
    'Customer_ID': [101, 102, 103, 104, 105, 105, 106, 107, 108, 109, 110, 111, 112, 113, 114],
    'Name': ['Ada', 'Tunde', 'Chioma', 'Musa', 'Funke', 'Funke', 'Emeka', 'Aisha', None, 'Bola', 'Chidi', 'Ngozi', 'Segun', 'Yemi', 'Kemi'],
    'Gender': ['Female', 'Male', 'female', 'M', 'F', 'F', 'Male', 'Female', 'Male', 'female', 'Male', 'F', 'M', 'Male', 'Female'],
    'Age': [25, 30, None, 45, 22, 22, 300, 28, 35, 29, None, 31, 27, 40, 26],
    'Join_Date': ['2024-01-15', '15/02/2024', '2024-03-10', '2024/04/05', '2024-05-20', '2024-05-20', '2024-06-01', '01-07-2024', '2024-08-12', '2024-09-03', '2024-10-11', '2024-11-05', '2024-12-01', '2025-01-10', '2025-02-14'],
    'Amount_Spent': [5000, 12000, 7500, None, 9000, 9000, 15000, 8000, 200000, 6500, 11000, 9500, 7000, 13000, None]
}

df = pd.DataFrame(data)
print(df)
```

```python
print("Shape:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
print("\nDuplicates:", df.duplicated().sum())
print("\nData types:\n", df.dtypes)
print("\nGender values:\n", df['Gender'].value_counts())
print("\nAge describe:\n", df['Age'].describe())
print("\nAmount describe:\n", df['Amount_Spent'].describe())
```

```python
# 1. Remove duplicates - keep first Funke, drop the second
df = df.drop_duplicates()
print("After removing duplicates:", df.shape)

# 2. Missing data
# Name: only 1 missing, we can't guess a name, so drop that row
# Age: fill with median (median is not affected by the 300 outlier)
# Amount_Spent: fill with median (median is not affected by the 200000 outlier)
df = df.dropna(subset=['Name'])
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Amount_Spent'] = df['Amount_Spent'].fillna(df['Amount_Spent'].median())

# 3. Standardise Gender: F, female -> Female, M, male -> Male
df['Gender'] = df['Gender'].replace({'F': 'Female', 'female': 'Female', 'M': 'Male', 'male': 'Male'})

# 4. Fix data types: Join_Date to datetime
df['Join_Date'] = pd.to_datetime(df['Join_Date'], dayfirst=True, format='mixed')
df['Customer_ID'] = df['Customer_ID'].astype(str)

print(df)
print("\nMissing after cleaning:\n", df.isnull().sum())
print("\nGender after cleaning:\n", df['Gender'].value_counts())
print("\nDtypes after cleaning:\n", df.dtypes)
```

```python
# 5. Outlier Detection with IQR for Age
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"Lower bound: {lower}, Upper bound: {upper}")
print("\nOutliers:\n", df[(df['Age'] < lower) | (df['Age'] > upper)])

# Decision: Cap the outlier at the upper bound, don't delete the customer
df['Age'] = df['Age'].clip(lower=lower, upper=upper)

print("\nFinal Clean Data:")
print(df)
print("\nFinal shape:", df.shape)
print(df.describe())
```

```python
df.to_csv('clean_data.csv', index=False)
print("Saved!")
print("Final shape:", df.shape)
print(df)
```
